```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_parquet('data/clean/2009-01-02-SPY.P-trade.parquet')
df.head()
```

```python
spread = df['spread']
spread_hourly = spread.resample('min').mean()
plt.plot(spread_hourly)
plt.title('Spread over Time')
plt.xlabel('Time')
plt.ylabel('Spread')
plt.show()
```
