```python
import pandas as pd
```

```python
df = pd.read_parquet('data/raw/SPY.P_events_short.parquet')

print(df.head())
```

```python
#to_datetime 
df['YYYY-MM-DDTHH:MM:SS.sss'] = pd.to_datetime(df['index'], unit='s')
df = df.set_index('YYYY-MM-DDTHH:MM:SS.sss')
df = df.drop(columns=['index', 'date'])

print(df.head())
```

```python
df['mid-price'] = 0.5 * (df['bid-price'] + df['ask-price'])
```
