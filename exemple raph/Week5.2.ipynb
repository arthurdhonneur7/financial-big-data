```python
import pandas as pd
```

```python
df = pd.read_parquet('data/raw/TRTH/equity/US/trade/SPY.P/2009-01-02-SPY.P-trade.parquet')

print(df.head())
```

```python
df = df.rename(columns={
    "xltime": "timestamp",
    "trade-price": "price",
    "trade-volume": "volume",
    "trade-stringflag": "flags",
    "trade-rawflag": "rawflag"
})

df['timestamp'] = pd.to_datetime(df['timestamp'], unit='D', origin='1899-12-30')
df = df.set_index('timestamp')

print(df.head())
```

```python
# Timzone :
# first date were market is open with flags
df["date"] = df.index.date
df["hour"] = df.index.hour

hours = df.groupby("date")["hour"].agg(["min", "max"])
print(hours.head())

open_df = df[df["flags"].str.contains("marketopen", case=False, na=False)]
open_df["hour"] = open_df.index.hour
open_df["date"] = open_df.index.date

hours = open_df.groupby("date")["hour"].agg(["min", "max"])
print(hours.head())
```
