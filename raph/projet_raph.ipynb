```python
import pandas as pd
import polars as pl
from pathlib import Path
```

```python
data_dir = Path(r"..\data\AAPL.OQ\AAPL.OQ")   # adapte si besoin

# on lit tous les fichiers .parquet et on les concatène
dfs = [pl.read_parquet(path) for path in data_dir.glob("*.parquet")]
df = pl.concat(dfs, how="vertical")

print(df.shape)
print(df.head())
```

```python
df = df.drop(["trade-stringflag", "trade-rawflag"])
```

```python
df = df.rename({
    "V1": "timestamp",
    "trade-price": "price",
    "trade-volume": "volume",
})
```
