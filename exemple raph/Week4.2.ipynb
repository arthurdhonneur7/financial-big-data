```python
import pandas as pd
import re
```

```python
df = pd.read_csv("data/raw/flights14.csv")
print(df.head())
```

```python
carr_nb = df.groupby("carrier").size()
print(carr_nb.head())
```

```python
dep_delay_mean = df.groupby("origin")["dep_delay"].mean()
arr_delay_mean = df.groupby("origin")["arr_delay"].mean()
delay_mean = 0.5*(dep_delay_mean + arr_delay_mean)
print(delay_mean)
```

```python
dep_delay_mean_2 = df.groupby(["origin", "dest"])["dep_delay"].mean()
arr_delay_mean_2 = df.groupby(["origin", "dest"])["arr_delay"].mean()
delay_mean_2 = 0.5*(dep_delay_mean_2 + arr_delay_mean_2)
print(delay_mean_2.head())
```
